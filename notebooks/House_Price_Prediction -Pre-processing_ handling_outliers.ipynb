# Outlier Detection and Handling for House Price Data

## Objective
The goal of this notebook is to identify and handle outliers in the `house_price_part1.csv` dataset. 

## Methodology: Winsorization (Capping)
In real estate datasets, features like lot area, living area, and sale price are naturally right-skewed. 
- **Dropping outliers** is avoided because extreme values (e.g., mansions) represent valid, valuable market data.
- **Imputation (Mean/Median)** is avoided because it artificially reduces variance and distorts the true distribution.
- **Log Transformation** changes the scale, which can complicate interpretation if not strictly required.

Therefore, the **"best method"** applied here is **Winsorization (Capping)**. We will cap the extreme values at the **1st and 99th percentiles**. This approach retains all data points, prevents extreme outliers from disproportionately skewing the data, and preserves the original scale of the features.

```python


# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings

# Ignore wanings
warnings.filterwarnings("ignore")

# Set visualization style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Load the dataset
df = pd.read_csv('../datasets/housing_no_missing_values.csv')

# drop the Unnamed column
#df.drop(columns="Unnamed: 0", inplace = True)

# Display basic info
print("Dataset Shape:", df.shape)
display(df.head())
```

```python
df.isna().sum().sum()
```

## 1. Initial Data Exploration & Numerical Column Selection
We need to isolate the numerical columns, as outlier detection methods like IQR and percentiles apply only to continuous/numerical data.

```python
# Select only numerical columns
num_cols = df.select_dtypes(include=[np.number]).columns

# drop the Order and the PID from numerical columns as they are not relevant for outlier detection
num_cols = num_cols.drop(['Order', 'PID'], errors='ignore')
print("Numerical Columns (after dropping Order and PID):", list(num_cols))
print(f"Number of numerical columns: {len(num_cols)}")


# Basic statistical summary to spot initial extreme values
display(df[num_cols].describe())
```

## 2. Visualizing Outliers (Before Treatment)
Before applying any mathematical treatment, we visualize the distributions and boxplots of the numerical features to visually confirm the presence and severity of outliers.

```python
# Plot distributions and boxplots for all numerical columns
fig, axes = plt.subplots(len(num_cols), 2, figsize=(15, 4 * len(num_cols)))
fig.suptitle('Distributions and Boxplots of Numerical Features (Before Treatment)', fontsize=16)

for i, col in enumerate(num_cols):
    # Histogram / KDE
    sns.histplot(df[col], kde=True, ax=axes[i, 0], color='skyblue')
    axes[i, 0].set_title(f'Distribution of {col}')
    axes[i, 0].set_xlabel('')
    
    # Boxplot
    sns.boxplot(y=df[col], ax=axes[i, 1], color='lightgreen')
    axes[i, 1].set_title(f'Boxplot of {col}')
    axes[i, 1].set_ylabel('')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

## 3. Handling Outliers using Winsorization (Capping)
We will define a function to cap the outliers. Instead of the strict 1.5 * IQR rule (which can be overly aggressive on naturally skewed real estate data), we will use the **1st and 99th percentiles**. This ensures we only trim the absolute most extreme anomalies while keeping the natural variance of the housing market intact.

```python
def winsorize_outliers(series, lower_percentile=0.01, upper_percentile=0.99):
    """
    Caps outliers at the specified percentiles (Winsorization).
    """
    lower_limit = series.quantile(lower_percentile)
    upper_limit = series.quantile(upper_percentile)
    
    # Clip the values to the limits
    return series.clip(lower_limit, upper_limit)

# Create a copy of the dataframe to apply the transformations
df_cleaned = df.copy()

# Apply the capping function to all numerical columns
for col in num_cols:
    df_cleaned[col] = winsorize_outliers(df_cleaned[col])

print("Outlier capping applied successfully to all numerical columns.")
```

## 4. Verification: Comparing Before and After
We will compare the statistical summaries and visualizations before and after the treatment to ensure the outliers have been successfully neutralized without destroying the data's core distribution.

```python
# Compare descriptive statistics
comparison = pd.DataFrame({
    'Original_Max': df[num_cols].max(),
    'Cleaned_Max': df_cleaned[num_cols].max(),
    'Original_Min': df[num_cols].min(),
    'Cleaned_Min': df_cleaned[num_cols].min(),
    'Original_Mean': df[num_cols].mean(),
    'Cleaned_Mean': df_cleaned[num_cols].mean()
})

# Calculate the percentage change in mean to ensure we didn't distort the center
comparison['Mean_Shift_%'] = ((comparison['Cleaned_Mean'] - comparison['Original_Mean']) / comparison['Original_Mean'] * 100).round(2)

display(comparison)
```

```python
# Visual verification: Boxplots After Treatment
fig, axes = plt.subplots(len(num_cols), 1, figsize=(10, 3 * len(num_cols)))
fig.suptitle('Boxplots of Numerical Features (After Winsorization)', fontsize=16)

# Handle case where there is only one numerical column
if len(num_cols) == 1:
    axes = [axes]

for i, col in enumerate(num_cols):
    sns.boxplot(y=df_cleaned[col], ax=axes[i], color='lightcoral')
    axes[i].set_title(f'Boxplot of {col} (Cleaned)')
    axes[i].set_ylabel('')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
